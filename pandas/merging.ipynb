# merging - joining - concatination


```python
import numpy as np
import pandas as pd
```

```python
employees = pd.DataFrame({
  "emp_id":[1,2,3,4,5],
  "emp_name":["kaif","ayaan","gautham","tejaswi","pranay"],
  "dept":["IT","architecture","business","designing","HR"]
})

salaries = pd.DataFrame({
  "emp_id":[1,2,3,8,10],
  "salary":[45000,120000,70000,37000,60000],
  "bonus" : [10000,2000,3000,4500,7000] 
})
```

```python
employees
```

```python
df = pd.merge(employees,salaries,how="inner")
df
```

```python
df = pd.merge(employees,salaries,how="right")
df
```

```python
df1 = pd.DataFrame({
  "A":["A0","A1","A2"],
  "B":["B0","B1","B2"],
  "C":["C0","C1","C2"]
})

df2 = pd.DataFrame({
  "A":["A3","A4","A5"],
  "B":["B3","B4","B5"],
  "C":["C3","C4","C5"]
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],axis=1)
```
